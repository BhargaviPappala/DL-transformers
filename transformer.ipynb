```python
# Attention Mechanisum

import numpy as np
#Input data
words=["The","cat","sat","on","mat"]
att_scores=np.array([0.1,0.4,0.8,0.2,0.3])  # manually assigned values
print("Original scores:")
print(att_scores)
exp_score=np.exp(att_scores)
attention_weights=exp_score / np.sum(exp_score)
print("\n Attention weights")
print(attention_weights)

# Show Importance

print("\n Word Importance")

for words, weight in zip(words,attention_weights):
    print(words, "->", round(weight,3))
```

```python
# Encoading - Decoading
input_sen=["I","am","happy"]
print("\n Input Sentance")
print(input_sen)

#Encoding
def encoading(sentance):
    encoaded=[]
    for word in sentance:
        encoaded.append(len(word))
    return encoaded
encoded_output=encoading(input_sen)
print("\n Encoded Output")
print(encoded_output)

#Decoading

def decoader(encoded_data):
    disctionary={1: "Je",
                2: "sius",
                5: "hear"}
    output=[]
    for value in  encoded_data:
        output.append(disctionary.get(value,"Unknow"))
    return output
decoaded_output=decoader(encoded_output)
print("\n Decoaded Output")
print(decoaded_output)
```

```python
#Positional encoading

import numpy as np
input_words=["I","love","Machine","Learning"]
embedding=np.array([[1,2,3,4],[2,3,4,5],[3,4,5,6],[4,5,6,7]])
print("\n Word Embedding")
print(embedding)

#Positional Matiix

position_encoded=np.zeros((4,4))
for pos in range(4):
    for i in range(4):
        if i%2==0:
            position_encoded[pos,i]=np.sin(pos/(10000**(i/4)))
        else:
            position_encoded[pos,i]=np.cos(pos/(10000**((i-1)/4)))
print("\n positional encoded matrix")
print(position_encoded)

#add with positional encoding

final_embedding=embedding + position_encoded
print("\n Final Embedding")
print(np.round(final_embedding,3))
```
